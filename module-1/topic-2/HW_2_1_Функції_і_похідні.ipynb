1. Задана функція залежності оцінки від кількості годин, витрачених на навчання: $ f(x) = 10x - 5 $.
  1. З допомогою Python обчисліть похідну цієї функції.
  2. Використовуйте бібліотеку `matplotlib` для побудови графіків функції та її похідної в діапазоні значень від 0 до 10 годин. Проаналізуйте, як зміна кількості годин вивчення впливає на оцінку.



```python
import matplotlib.pyplot as plt
import numpy as np
import sympy as sp
```

```python
# Визначення функції та її похідної
x = np.linspace(0, 10, 50)
f = 10*x - 5

# Похідна f'(x) = 10. Для коректного відображення на графіку
# необхідно, щоб df мав таку ж розмірність, як x.
df = np.full_like(x, 10)

# Створення графіку
plt.figure(figsize=(10, 5))
plt.plot(x, f, label='f(x) = 10*x - 5', color='blue')
plt.plot(x, df, label="f'(x) = 10", color='green')


# Додавання легенди, міток та заголовка
plt.title("Графік функції та її похідної")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)

plt.show()
```

2. Використовуючи Python і `SymPy`, обчисліть похідні для функцій $ f(x) = \sin(x) $ та $ f(x) = \ln(x) $. Зобразіть ці функції та їхні похідні за допомогою `matplotlib`.



```python
# Визначаємо символічну змінну
x = sp.symbols('x')

# Функція сінуса
f_sin = sp.sin(x)
derivative_f_sin = sp.diff(f_sin, x)

# Функція логарифмічна
f_log = sp.log(x)
derivative_f_log = sp.diff(f_log, x)

print("Похідна функція сінуса sin(x):", derivative_f_sin)
print("Похідна логарифмічної функції log(x):", derivative_f_log)
```

```python
# масив точок Х для графіків
x_arr = np.linspace(0, 10, 50)
```

```python
# Визначаємо числові представлення функцій SymPy для графіків:
f_sin_np = sp.lambdify(x, f_sin, 'numpy')
derivative_f_sin_np = sp.lambdify(x, derivative_f_sin, 'numpy')

# Розраховуємо значення функцій по x_arr
y_f_sin = f_sin_np(x_arr)
y_derivative_f_sin = derivative_f_sin_np(x_arr)

# Графік
plt.figure(figsize=(10, 5))
plt.plot(x_arr, y_f_sin, label='f(x) = sin(x)', color='blue')
plt.plot(x_arr, y_derivative_f_sin, label="f'(x) = cos(x)", color='green')

plt.title("Графік функції синуса та її похідної")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)

plt.show()
```

```python
# масив точок Х для графіків
x_arr = np.linspace(1, 10, 50)

# Визначаємо числові представлення функцій SymPy для графіків:
f_log_np = sp.lambdify(x, f_log, 'numpy')
derivative_f_log_np = sp.lambdify(x, derivative_f_log, 'numpy')

# Розраховуємо значення функцій по x_arr
y_f_log = f_log_np(x_arr)
y_derivative_f_log = derivative_f_log_np(x_arr)

# Графік
plt.figure(figsize=(10, 5))
plt.plot(x_arr, y_f_log, label='f(x) = log(x)', color='blue')
plt.plot(x_arr, y_derivative_f_log, label="f'(x) = 1/x", color='green')

plt.title("Графік функції логарифма та його похідної")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)

plt.show()
```


3. Загальні витрати компанії від обсягу виробництва задані функцією $ f(x) = x^2 - 5x + 1 $, де $ x $ — кількість вироблених одиниць. Обчисліть похідну цієї функції з допомогою SymPy та візуалізуйте значення функції і її похідної на відрізку від 0 до 100.

```python
x_arr_custom = np.linspace(0, 100, 100)

# Визначаємо символічну змінну
x = sp.symbols('x')

# Функція f(x) = x^2 - 5x + 1
f_custom = x**2 -5*x + 1
derivative_f_custom = sp.diff(f_custom, x)

print("Похідна функції x^2 - 5x + 1:", derivative_f_custom)

# Визначаємо числові представлення функцій SymPy для графіків:
f_custom_np = sp.lambdify(x, f_custom, 'numpy')
derivative_f_custom_np = sp.lambdify(x, derivative_f_custom, 'numpy')

# Розраховуємо значення функцій по x_arr_custom
y_f_custom = f_custom_np(x_arr_custom)
y_derivative_f_custom = derivative_f_custom_np(x_arr_custom)

# Графік
plt.figure(figsize=(10, 5))
plt.plot(x_arr_custom, y_f_custom, label='f(x) = x^2 - 5x + 1', color='blue')
plt.plot(x_arr_custom, y_derivative_f_custom, label="f'(x) = 2x - 5", color='green')

# Add legend, labels, and title
plt.title("Графік функції та її похідної")
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True)

plt.show()
```
