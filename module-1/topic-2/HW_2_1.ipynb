### Завдання 1: Z-тест для однієї вибірки

**Мета**: Перевірити, чи середнє значення вибірки відрізняється від гіпотетичного середнього.

**Припущення**: Середній ріст студентів університету вважається 170 см. Ми хочемо перевірити, чи середній ріст випадкової вибірки студентів істотно більший. Вибірка задана в змінній `heights`.

Для виконання завдання імпортуйте метод для проведення `ztest`, виконайте Z-тест з Python, виведіть p-value і зробіть заключення, чи справдилось припущення про середній зріст студентів на рівні значущості 0.05?

Використовуємо тут Z-тест, оскільки вибірка достатньо велика (більше 30 екземплярів) і можемо вважати, що її стандартне відхилення близьке до ст. відхилення генеральної сукупності, а значить - відоме.

```python
heights = [174, 171, 175, 179, 170, 170, 179, 175, 169, 174, 169, 169, 173, 162, 163, 169, 166,
           173, 167, 164, 179, 170, 172, 164, 169, 175, 169, 174, 169, 169, 173, 162, 177]
```

```python
import numpy as np
from statsmodels.stats.weightstats import ztest, ttest_ind
```

```python
null_mean = 170
alpha = 0.05
```

```python
# перевірка результатів
print("Середнє вибірки: {:.2f}".format(np.mean(heights)))
print("Стандартне відхилення вибірки: {:.2f}".format(np.std(heights)))
```

```python
# Наша гіпотеза H0 полягає в тому, що середній ріст виспадкової вибірки студентів дорівнює середньому зросту студентів університету 170см.
# H1 гіпотеза - що середній ріст випадкової вибірки істотно більший
ztest_Score, p_value = ztest(heights, value = null_mean, alternative='larger')
```

```python
ztest_Score, p_value
```

```python
if(p_value <  alpha):
    print("Відхилити Н0.")
else:
    print("Н0 не може бути відхилена.")
```

### **Висновок:**
Н0 не може бути відхилена. Середній ріст виспадкової вибірки студентів дорівнює середньому зросту студентів університету 170см.

### Завдання 2: Z-тест для двох незалежних вибірок

**Завдання 2.1.**

**Мета**: Перевірити, чи існує статистична різниця між середніми оцінками двох груп студентів.

**Припущення**: Група A отримала новий навчальний курс, а група B продовжувала зі стандартним курсом. Ми перевіряємо, чи є новий курс ефективнішим.

Виконайте Z-тест з Python, виведіть p-value і зробіть заключення, чи справдилось припущення щодо студентів на рівні значущості 0.05?

```python
group_a_scores = [78.55, 72.25, 79.88, 75.  , 76.54, 74.99, 87.26, 77.93, 72.71,
       82.11, 71.9 , 79.04, 68.2 , 71.36, 78.98, 81.69, 78.86, 77.42,
       76.49, 70.61, 74.4 , 75.7 , 83.29, 79.72, 69.18, 79.62, 76.07,
       74.62, 81.06, 83.15, 82.66, 73.8 , 76.45, 79.66, 82.88, 75.6 ,
       77.07, 72.47, 72.02, 82.06]

group_b_scores = [81.78, 74.64, 80.02, 76.81, 71.77, 76.81, 82.69, 74.82, 82.82,
       61.9 , 79.11, 75.44, 73.5 , 75.46, 65.06, 73.9 , 76.79, 82.39,
       72.41, 70.96, 72.49, 79.58, 76.64, 72.35, 77.57, 75.49, 79.84,
       71.49, 73.36, 73.04, 67.68, 76.48, 76.31, 75.03, 73.83, 67.92,
       72.9 , 73.29, 70.99, 74.19]
```

```python
print("Середнє вибірки group_a_scores: {:.2f}".format(np.mean(group_a_scores)))
print("Стандартне відхилення вибірки group_a_scores: {:.2f}".format(np.std(group_a_scores)))

print("Середнє вибірки group_b_scores: {:.2f}".format(np.mean(group_b_scores)))
print("Стандартне відхилення вибірки group_b_scores: {:.2f}".format(np.std(group_b_scores)))
```

```python
# Гіпотеза H0 - зміна курсі не вплинула на середні оцінки
# Гіпотеза H1 - зміна курсу для групи А покращила резльтати, а отже результати для групи B гірші (тому обирамо left-tailed z test)

ztest_Score_a_b, p_value_a_b = ztest(group_a_scores, group_b_scores, alternative='smaller')

print('p_value_a_b =', p_value_a_b)

if(p_value_a_b <  alpha):
    print("Відхилити Н0.")
else:
    print("Н0 не може бути відхилена. Зміна курсу не вплинула достатньо на оцінки.")
```

### **Висновок:**
Н0 не може бути відхилена. Зміна курсу не вплинула достатньо на оцінки.

### Завдання 3\*: Z-тест для двох **пов'язаних** вибірок

**Мета**: Перевірити, чи вплинув тренінг на продуктивність працівників, коли всі справробітники проходили тренінг.

**Припущення**: Наші працівники проходять однаковий тренінг, і ми хочемо з'ясувати, чи покращилися їхні показники після тренінгу на рівні значущості 0.05. Виконайте Z-тест з Python, виведіть p-value і зробіть заключення, чи поліпшились показники працівників на рівні значущості 0.05?

Зверніть увагу, ці вибірки повʼязані між собою, значить не буде коректно виконати z-test між ними, натомість, ми можемо порівняти, чи різниця в результатах (продуктивності співробітників) є стат значущою.

```python
before_training = [57.82, 37.63, 36.8 , 55.22, 52.97, 52.5 , 53.46, 43.2 , 52.32,
       52.93, 42.86, 68.66, 54.74, 38.09, 56.57, 40.25, 57.87, 61.59,
       41.79, 59.63, 54.13, 58.22, 68.97, 47.55, 42.46, 41.1 , 41.84,
       49.23, 53.41, 52.77]

after_training = [62.47, 40.66, 42.7 , 57.69, 61.41, 56.76, 54.75, 44.06, 56.29,
       55.48, 47.28, 72.6 , 57.59, 39.39, 56.54, 42.36, 62.58, 65.01,
       42.3 , 62.98, 57.9 , 59.45, 72.28, 50.66, 43.18, 44.82, 45.96,
       54.4 , 58.52, 53.01]
```

```python
# так як вибірки пов"язані ми не можемо використати їх для z-тесту, хоча розмір вибірки підходить для тесту. Але, ми можемо привести задачу до однієї вибірки - вибірці наскільки змінились результати,
# тобто працювати з 1 числом для кожного елементу вибірок, де це число є різницею між значенням після тренінгу і до:

diff = np.array(after_training) - np.array(before_training)
diff, len(diff)
```

```python
# тепер ми можемо будувати гіпотези для вибірки оцих різниць
# H0 - після проходження тренінгу змін не відбулось (тобто середнє значення = 0, ніяких змін)
# H1 - після проходження тренінгу результати покращились (ми бачимо і візуально з масву, але проведемо тест)

# 2. One-sample Z-test
z_stat, p_value = ztest(
    diff,
    value=0,
    alternative="larger"  # H1: mean(diff) > 0
)

z_stat, p_value
ztest_Score_for_diff, p_value_for_diff = ztest(diff, value = 0, alternative='larger') # H1: mean(diff) > 0

print('p_value = ', p_value_for_diff)
# e-20 - це дуууууже маленьке число
print(f"p_value is {p_value_for_diff:.4f}")
```

```python
if(p_value_for_diff <  alpha):
    print("Відхилити Н0. Після проходження тренінгу результати покращились")
else:
    print("Н0 не може бути відхилена. Після проходження тренінгу змін не відбулось.")
```

### **Висновок:**

хоча я не могла на початку для двох залежних вибірок застосувати z тест, але при приведенні до однієї вибірки проблема була вирішена.

Як показав тест - результати покращились після проходження тренінгу:
*p_value < alpha*
